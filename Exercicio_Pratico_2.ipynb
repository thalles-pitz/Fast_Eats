# Parte 1 — Preparação do ambiente



```python
import pandas as pd
df = pd.read_csv('/content/qualidade_ar_brasil.csv')
```

# Parte 2 — Inspeção inicial do dataset

```python
df.shape
df.head()
df.info()
df.describe()
df.isnull().sum()
```

1. 800 medições no dataset.
2. 10 cidades: Brasília, Curitiba, Salvador, Belo Horizonte, Recife, Fortaleza, Rio de Janeiro, Porto Alegre, Manaus, São Paulo.
3. Poluentes: pm25, pm10, no2, o3.
4. Não há valores ausentes (todas as colunas com 800 registros não nulos).

# Parte 3 — Análise por cidade

```python
df['cidade'].unique()
```

```python
df.groupby('cidade')[['pm25','pm10','no2','o3']].mean()
```

1. Maior média de PM2.5: São Paulo (≈33,7).
2. Maior média de NO2: São Paulo (≈39,3).
3. Sim — São Paulo aparece consistentemente no topo em quase todos os poluentes (PM2.5, PM10, NO2), indicando poluição estruturalmente mais alta.

# Parte 4 — Análise do índice de qualidade do ar

```python
df.groupby('cidade')['indice_qualidade_ar'].mean().sort_values(ascending=False)
```

1. Pior qualidade média: São Paulo (59,3).
2. Melhor qualidade média: Salvador (42,3).

# Parte 5 — Filtragem por nível de alerta

```python
df_alerta = df[df['nivel_alerta'].isin(['Ruim','Muito Ruim'])]
df_alerta.head()
```

1. 14 registros classificados como Ruim ou Muito Ruim.
2. Maior frequência: São Paulo (8 ocorrências), seguida por Rio de Janeiro (2) e Brasília (2).

# Parte 6 — Ranking dos dias mais críticos

```python
df.sort_values('indice_qualidade_ar', ascending=False).head(10)
```

1. Pior dia: 19/08/2025, índice 86,3.
2. Cidade: São Paulo.
3. Poluentes mais altos nesse dia: PM10 (78,8) e NO2 (55,0), com PM2.5 também elevado (55,2).

# Parte 7 — Análise temporal

```python
df['data'] = pd.to_datetime(df['data'])
```

```python
df['mes'] = df['data'].dt.month
```

```python
df.groupby('mes')['indice_qualidade_ar'].mean()
```

1. Meses com pior qualidade: junho (52,9), julho (51,3) e setembro (49,9) — período de inverno/seca no Sudeste/Centro-Oeste.
2. Padrão sazonal: sim — piora no inverno (menor umidade, mais dispersão de poluentes) e melhora no verão (fevereiro tem o melhor índice, 43,0).

# Parte 8 — Correlação entre variáveis

```python
df[['temperatura','pm25','pm10','no2','o3']].corr()
```

1. Correlação temperatura x poluição: fraca e negativa (ex.: pm25 = -0,11; no2 = -0,17), sugerindo que temperaturas mais baixas acompanham leve aumento da poluição.
2. Poluentes mais relacionados entre si: PM2.5 e PM10 (correlação ≈0,25), a mais forte do grupo, ainda que moderada.

1. São Paulo apresenta a pior qualidade média do ar do país.

3. Sim, os meses de junho a setembro (inverno) são mais problemáticos.
4. Cidades que merecem maior atenção: São Paulo (prioridade máxima), seguida de Brasília e Rio de Janeiro, que também tiveram alertas críticos e médias elevadas.
